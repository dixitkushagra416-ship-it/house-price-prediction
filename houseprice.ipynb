```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
house_data=pd.read_csv("HousePricePrediction.csv")
house_data
sns.scatterplot(
    data=house_data,
    x="LotArea",
    y="SalePrice"
)
```

```python
house_data=house_data.drop(columns=["MSZoning","LotConfig","BldgType","Exterior1st"])
```

```python
house_data["SalePrice"].isnull().sum()
house_data = house_data.dropna(subset=["SalePrice"])
```

```python
house_data["SalePrice"].isnull().sum()
```

```python
x=house_data.drop(columns=["SalePrice"])
y=house_data["SalePrice"]
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test, y_pred)
n=x_test.shape[0]
p=x_test.shape[1]
adjusted_r2=1-((1-r2)*(n-1)/(n-p-1))
```
